```python
import sys
from collections import Counter

sys.path.insert(0, "../src")

from data.ingest import generate_synthetic
from data.availability import SIMULATED

users, restaurants, interactions = generate_synthetic()
print("availability simulated:", SIMULATED)
print(len(users), "users,", len(restaurants), "restaurants,", len(interactions), "events")
print(Counter(e.event_type for e in interactions))
```

```python
from collections import Counter

print("cuisines", Counter(r.cuisine for r in restaurants))
print("price levels", Counter(r.price_level for r in restaurants))
print("mean rating", sum(r.rating for r in restaurants) / len(restaurants))
```
